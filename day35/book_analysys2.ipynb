```python
with open("miracle-in-the-andes.txt", "r", encoding="utf-8") as file:
    book = file.read()
```

# The most used words (non-articles)

```python
import re
pattern = re.compile("[a-zA-Z]+")
findings = re.findall(pattern, book.lower())
```

```python
d = {}
for word in findings:
    if word in d.keys():
        d[word] = d[word] + 1
    else:
        d[word] = 1
```

```python
d_list = [(value, key) for key, value in d.items()]
d_list = sorted(d_list, reverse=True)
d_list[:5]
```

```python
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
english_stopwords = stopwords.words("english")
english_stopwords
```

```python
filtered_words = []
for count, word in d_list:
    if word not in english_stopwords:
        filtered_words.append((word, count))
```

# Sentiment Analisis: What is the most positive and the most negative chapter?

### An example

```python
nltk.download('vader_lexicon')
```

```python
from nltk.sentiment import SentimentIntensityAnalyzer
```

```python
# analyzer = SentimentIntensityAnalyzer
analyzer = SentimentIntensityAnalyzer()
```

```python
scores = analyzer.polarity_scores(text="Hey, look how beutifull the trees are. I love them.")
```

```python
if scores["pos"] > scores["neg"]:
    print("It's a positive text!")
else:
    print("It's a negative text!")
```

```python
scores = analyzer.polarity_scores(book)
scores
```

### Chapters sentiment

```python
import re
pattern = re.compile("Chapter [0-9]+")
chapters = re.split(pattern, book)
```

```python
chapters = chapters[1:]
```

```python
for number, chapter in enumerate(chapters):
    scores = analyzer.polarity_scores(chapter)
    print(number + 1, scores)
```
